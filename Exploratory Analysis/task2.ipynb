```python
import pandas as pd
import numpy as np

df = pd.read_csv("employees.csv")

print("First 5 Rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Select numeric columns
numeric_columns = ["Age", "Years of Experience", "Salary"]

# Calculate summary statistics
summary = df[numeric_columns].agg([
    "mean",
    "median",
    "min",
    "max",
    "std"
])

print("Employee Summary Statistics:")
display(summary)
```

```python
# Compare employees by gender
gender_analysis = df.groupby("Gender").agg(
    Employee_Count=("Salary", "count"),
    Average_Salary=("Salary", "mean"),
    Median_Salary=("Salary", "median"),
    Minimum_Salary=("Salary", "min"),
    Maximum_Salary=("Salary", "max")
)

print("Gender-wise Salary Analysis:")
display(gender_analysis.round(2))
```

```python
# Compare employees by education level
education_analysis = df.groupby("Education Level").agg(
    Employee_Count=("Salary", "count"),
    Average_Salary=("Salary", "mean"),
    Median_Salary=("Salary", "median"),
    Average_Experience=("Years of Experience", "mean")
)

print("Education-wise Salary Analysis:")
display(education_analysis.round(2))
```

```python
# Calculate correlation between numeric columns
correlation = df[["Age", "Years of Experience", "Salary"]].corr()

print("Correlation Matrix:")
display(correlation.round(3))

# Find the top 3 relationships
correlation_pairs = correlation.unstack()

# Remove duplicate pairs and self-correlations
correlation_pairs = correlation_pairs[
    correlation_pairs.index.get_level_values(0) <
    correlation_pairs.index.get_level_values(1)
]

# Sort by absolute correlation
top_3 = correlation_pairs.abs().sort_values(ascending=False).head(3)

print("\nTop 3 Correlation Relationships:")
display(top_3)
```

```python
# Business Question 1: What is the average employee salary?
print("Q1: What is the average employee salary?")
print("Average Salary:", round(df["Salary"].mean(), 2))


# Business Question 2: Which gender has the highest average salary?
print("\nQ2: Which gender has the highest average salary?")
gender_salary = df.groupby("Gender")["Salary"].mean()
print(gender_salary.round(2))
print("Highest:", gender_salary.idxmax())


# Business Question 3: Which education level has the highest average salary?
print("\nQ3: Which education level has the highest average salary?")
education_salary = df.groupby("Education Level")["Salary"].mean()
print(education_salary.round(2))
print("Highest:", education_salary.idxmax())


# Business Question 4: What is the relationship between experience and salary?
print("\nQ4: Does experience relate to salary?")
experience_corr = df["Years of Experience"].corr(df["Salary"])
print("Experience-Salary Correlation:", round(experience_corr, 3))


# Business Question 5: What is the minimum and maximum salary?
print("\nQ5: What is the salary range?")
print("Minimum Salary:", df["Salary"].min())
print("Maximum Salary:", df["Salary"].max())
```

```python
# Create a second table with sample training budgets
education_levels = sorted(df["Education Level"].dropna().unique())

education_budget = pd.DataFrame({
    "Education Level": education_levels,
    "Training_Budget": [
        10000 + (i * 2000) for i in range(len(education_levels))
    ]
})

print("Education Budget Table:")
display(education_budget)


# Merge employee data with the budget table
merged_df = pd.merge(
    df,
    education_budget,
    on="Education Level",
    how="left"
)

print("\nMerged Employee Data:")
display(merged_df.head(10))


# Calculate total training budget by education level
print("\nTotal Training Budget by Education Level:")
display(
    merged_df.groupby("Education Level").agg(
        Employee_Count=("Salary", "count"),
        Total_Training_Budget=("Training_Budget", "sum")
    )
)
```
